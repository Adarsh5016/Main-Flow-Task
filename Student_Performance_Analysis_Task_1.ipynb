# Data Analysis and Data Science Task - 1

**Dataset:** Student Performance Dataset (student-mat.csv)

**Objective:** Analyze student exam scores and uncover insights using Python.


## 1. Data Loading

```python
import pandas as pd

# Load the dataset
file_path = 'student-mat.csv'
df = pd.read_csv(file_path, sep=';')

# Display the first few rows
df.head()
```

## 2. Data Exploration

```python
# Check for missing values
df.isnull().sum()
```

```python
# Display data types
df.dtypes
```

```python
# Understand the dataset size
df.shape
```

## 3. Data Cleaning

```python
# Drop duplicates
df = df.drop_duplicates()

# If there were missing values, handle them here (not expected in this dataset)
# df = df.fillna(df.median())  # Example
```

```python
# Confirm cleaning
df.isnull().sum(), df.duplicated().sum()
```

## 4. Data Analysis Questions

**Q1. What is the average score in math (G3)?**

```python
average_g3 = df['G3'].mean()
average_g3
```

**Q2. How many students scored above 15 in G3?**

```python
above_15 = df[df['G3'] > 15].shape[0]
above_15
```

**Q3. Is there a correlation between study time and G3?**

```python
correlation = df['studytime'].corr(df['G3'])
correlation
```

**Q4. Which gender has a higher average final grade (G3)?**

```python
avg_by_gender = df.groupby('sex')['G3'].mean()
avg_by_gender
```

## 5. Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot a histogram of final grades
plt.figure(figsize=(8, 5))
sns.histplot(df['G3'], bins=10, kde=True)
plt.title('Distribution of Final Grades (G3)')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Scatter plot between study time and final grade
plt.figure(figsize=(8, 5))
sns.scatterplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grade (G3)')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')
plt.show()
```

```python
# Bar chart comparing average scores by gender
plt.figure(figsize=(8, 5))
sns.barplot(x='sex', y='G3', data=df, estimator='mean')
plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')
plt.show()
```

## Summary & Insights
- The average final grade (G3) provides a general idea of student performance.
- Students who scored above 15 are relatively fewer.
- A positive correlation between study time and grades indicates effort matters.
- Gender-wise performance can help in understanding learning trends.
